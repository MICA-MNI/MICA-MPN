```python
import os
import pandas as pd
def mk_participants_tsv(rawdata_dir):
    # make bids compliant participants.tsv from rawdata    

    subjects = sorted([d for d in os.listdir(rawdata_dir) if d.startswith('sub-')])
    participants_df = pd.DataFrame({'participant_id': subjects})

    # Save the DataFrame to participants.tsv in the rawdata directory
    tsv_path = os.path.join(rawdata_dir, 'participants.tsv')
    participants_df.to_csv(tsv_path, sep='\t', index=False)
    print(f"Made {tsv_path} with {len(subjects)} participants.")

mk_participants_tsv('/host/verges/tank/data/BIDS_MPN/rawdata/')
```
