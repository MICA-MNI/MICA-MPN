```python
from bids import BIDSLayout as bl
```

```python
root = '/host/verges/tank/data/BIDS_MPN/'
mpn = bl(root, validate=False)

inv1_files = mpn.get(
    suffix='inv-1_MP2RAGE',
    acquisition='05mm',
    extension=['.nii', '.nii.gz']
)
inv2_files = mpn.get(
    suffix='inv-2_MP2RAGE',
    acquisition='05mm',
    extension=['.nii', '.nii.gz']
)
uni_files = mpn.get(
    suffix='UNIT1',
    acquisition='05mm',
    extension=['.nii', '.nii.gz']
)

# Subjects with all three (UNIT1 + INV1 + INV2) at 0.5mm
sub_inv1 = {f.subject for f in inv1_files}
sub_inv2 = {f.subject for f in inv2_files}
sub_uni  = {f.subject for f in uni_files}

subjects_05mm = sub_inv1 & sub_inv2 & sub_uni

print(f"Number of subjects with 0.5mm MP2RAGE: {len(subjects_05mm)}")
print("Subjects:", sorted(subjects_05mm))
```
