```python
import os
from collections import defaultdict

root = "/host/verges/tank/data/BIDS_MPN/derivatives/micapipe_v0.2.0"

sessions_by_sub = defaultdict(set)

for dirpath, dirnames, filenames in os.walk(root):
    parts = dirpath.split(os.sep)
    # look for .../sub-XXX/ses-YYY/...
    for i, part in enumerate(parts):
        if part.startswith("sub-"):
            sub = part  # e.g. "sub-001"
            # check next part for ses- label
            if i + 1 < len(parts):
                next_part = parts[i + 1]
                if next_part.startswith("ses-"):
                    ses = next_part  # e.g. "ses-01"
                    sessions_by_sub[sub].add(ses)

# now report
for sub, sess in sorted(sessions_by_sub.items()):
    if len(sess) == 1:
        # only one session string: print just that
        unique_ses = next(iter(sess))
        print(unique_ses)
    else:
        # more than one: print subject and all sessions
        print(sub, sorted(sess))
```
